```python
def isUgly(n):
    if n == 1:
        return True
    if n <= 0:
        return False   
    while n != 1:
        if n % 2 == 0:
            n /= 2
        elif n % 5 == 0:
            n /= 5
        elif n % 3 == 0:
            n /= 3
        else:
            return False
    return True
```

```python
isUgly(6)
```

```python
def isUgly(n):
    factor = [2, 3, 5]
    if n <= 0:
        return False
    if n == 1:
        return True
    for i in factor:
        while n % i == 0:
            n /= i
    return n == 1
```
