```python
def generateMatrix(n):
    # res = [[0] * n] * n
    res = [[0] * n for _ in range(n)] 
    top = left = 0
    bot = right = n - 1
    num = 1
    while left <= right and top <= bot:
        for col in range(left, right + 1):
            res[top][col] = num
            num += 1
        top += 1
        if top > bot:
            break

        for row in range(top, bot + 1):
            res[row][right] = num
            num += 1
        right -= 1

        if left > right:
            break

        for col in range(right, left - 1, -1):
            res[bot][col] = num
            num += 1
        bot -= 1

        if top > bot:
            break

        for row in range(bot, top - 1, -1):
            res[row][left] = num
            num += 1
        left += 1
        if left > right:
            break
    return res
```

```python
generateMatrix(n)
```
