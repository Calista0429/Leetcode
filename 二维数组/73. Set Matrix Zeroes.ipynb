```python
def setZeroes(matrix):
    rows = len(matrix)
    cols = len(matrix[0])
    stg_row = [0] * rows
    stg_col = [0] * cols
    for i in range(rows):
        for j in range(cols):
            if matrix[i][j] == 0:
                stg_row[i] = 1
                stg_col[j] = 1

    for j in range(cols):
        if stg_col[j] == 1:
            cur_row = 0
            while cur_row != rows:
                matrix[cur_row][j] = 0
                cur_row += 1

    for i in range(rows):
        if stg_row[i] == 1:
            cur_col = 0
            while cur_col != cols:
                matrix[i][cur_col] = 0
                cur_col += 1
    return matrix
```
