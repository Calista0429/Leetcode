```python
"""
这道题使用当前位置的row和col来进行取数和边界控制，使用一个变量来决定是向上走还是向下走。如果是向上走，需要注意 当前行不要小于0（1情况）和列不要超过总列数（7->3情况）。
如果是从上向下遍历（2->4的情况）需要注意列不要小于0（2->4）并且行要小于总行数（6->8）
"""
```

```python
def findDiagonalOrder(mat):
    #总行数，列数
    rows, cols = len(mat), len(mat[0])
    #返回数组
    res = []
    #用于遍历当前位置
    cur_row = cur_col = 0
    #遍历顺序
    going_up = True
    while len(res) != rows * cols:
        if going_up:
            while cur_col < cols and cur_row >= 0:
                res.append(mat[cur_row][cur_col])
                cur_row -= 1
                cur_col += 1
            #这里的cur_col是从下标0开始
            if cur_col == cols:
                cur_col -= 1
                cur_row += 2
            else:
                cur_row += 1
            going_up = False
        else:
            while cur_row < rows and cur_col >= 0:
                res.append(mat[cur_row][cur_col])
                cur_col -= 1
                cur_row += 1
            if cur_row == rows:
                cur_row -= 1
                cur_col += 2
            else:
                cur_col += 1
            going_up = True
    return res
```

```python
mat = [[1,2,3],[4,5,6],[7,8,9]]
findDiagonalOrder(mat)
```
