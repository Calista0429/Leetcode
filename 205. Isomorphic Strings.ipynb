```python
s = "title"
t = "paper"
def isIsomorphic(s, t):
    char_map = {}
    for sc, tc in zip(s,t):
        if sc in char_map:
            if char_map[sc] != tc:
                return False
            elif tc not in char_map.values():
                return False
        char_map[sc] = tc
    return True
```

```python
isIsomorphic(s, t)
```
