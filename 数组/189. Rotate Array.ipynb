```python
def rotate(nums, k):
    k = k % len(nums)
    
    if k == 0:
        return nums
    
    rotate_num = nums[-k:]
    res = []
    
    for num in nums:
        if num not in rotate_num:
            res.append(num)
    return rotate_num + res
```

```python
nums = [1,2,3,4,5,6,7]
k = 3
rotate(nums, k)
```

```python
def rotate(nums, k):
    k = k % len(nums)
    
    if k == 0:
        return nums
    
    nums[:k], nums[-k:] = nums[-k:], nums[:k]
    return nums
```

```python
nums = [-1,-100,3,99]
k = 2

nums = [1,2,3,4,5,6,7]
k = 3
rotate(nums, k)
```
