```python
nums = [1,2,3,1,2,3]
k = 2
def containsNearbyDuplicate(nums, k):
    seen = {}
    for idx, num in enumerate(nums):
        if num in seen and idx - seen[num] <= k:
            return True
        else:
            seen[num] = idx
    return False
```

```python
containsNearbyDuplicate(nums, k)
```
