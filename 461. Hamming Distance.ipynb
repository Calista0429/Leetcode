```python
"""
The Hamming distance between two integers is the number of positions at which the corresponding bits are different.

Given two integers x and y, return the Hamming distance between them.

 

Example 1:

Input: x = 1, y = 4
Output: 2
Explanation:
1   (0 0 0 1)
4   (0 1 0 0)
       ↑   ↑
The above arrows point to positions where the corresponding bits are different.
Example 2:

Input: x = 3, y = 1
Output: 1
 
"""
```

```python
x = 1
y = 4
```

```python
bin(x)
```

```python
bin(y)
```

```python
bin_x = (len(bin(y)) - len(bin(x))) * '0' + bin(x)[2:]
```

```python
bin(y)[2:]
```

```python
def hammingDistance(x, y):
    yp, xp = 0, 0
    res = 0
    while yp < len(bin(y)[2:]) and xp < len(bin(y)):
        if bin(y)[2:][yp] != bin_x[xp]:
            res += 1
            yp += 1
            xp += 1
        else:
            yp += 1
            xp += 1
    return res
```
