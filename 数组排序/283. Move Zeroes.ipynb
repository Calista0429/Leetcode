```python
nums = [0,1,0,3,12]
```

```python
zeroCount = nums.count(0)
for c in range(zeroCount):
    nums.remove(0)
```

```python
nums = [0,1,0,3,12]
left = 0
for right in range(1, len(nums)):
    if nums[right] != 0:
        nums[left], nums[right] = nums[right], nums[left]
        left += 1
```
