```python
def get_next_number(n):
    output = 0
    while n:
        digit = n % 10
        output += digit ** 2
        n //= 10
    return output

def happy_number(n):
    visit = set()
    while n not in visit:
        visit.add(n)
        n = get_next_number(n)
        if n == 1:
            return True
    return False
```
