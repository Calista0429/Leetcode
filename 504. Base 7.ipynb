```python
"""
Given an integer num, return a string of its base 7 representation.

 

Example 1:

Input: num = 100
Output: "202"
Example 2:

Input: num = -7
Output: "-10"

"""
```

```python
res = ""
num = 100
sign = "-" if num < 0 else ""
num = abs(num)
while num:
    res = str(num % 7) + res
    num //= 7
res = sign + res
```
