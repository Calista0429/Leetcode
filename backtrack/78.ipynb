```python
def subsets(nums: list[int]) -> list[list[int]]:
    def backtracking(index, path):
        if path not in res:
            res.append(path[:])
            print(path[:])
            return
            
        for i in range(index, len(nums)):
            path.append(nums[i])
            backtracking(index, path)
            path.pop()
    res = []
    res.append([])
    backtracking(0, [])
```

```python
subsets(nums = [1,2,3])
```
