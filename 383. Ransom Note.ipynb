```python
"""
Given two strings ransomNote and magazine, return true if ransomNote can be constructed by using the letters from magazine and false otherwise.

Each letter in magazine can only be used once in ransomNote.

 

Example 1:

Input: ransomNote = "a", magazine = "b"
Output: false
Example 2:

Input: ransomNote = "aa", magazine = "ab"
Output: false
Example 3:

Input: ransomNote = "aa", magazine = "aab"
Output: true
 

"""
```

```python
ransomNote = "aa" 
magazine = "ab"
def canConstruct(ransomNote, magazine):
    mag_list = list(magazine)
    count = 0
    for char in ransomNote:
        if char in mag_list:
            count += 1
            mag_list.remove(char)
    if count == len(ransomNote):
        return True
    return False
```

```python
canConstruct(ransomNote, magazine)
```

```python
from collections import Counter
ransomNote = "aa" 
magazine = "ab"
def canConstruct(ransomNote, magazine):
    mag_count = Counter(magazine)
    note_count = Counter(ransomNote)
    count = 0
    print("mag_count:", mag_count, "note_count:", note_count)
    for char, total in note_count.items():
        if char in mag_count and mag_count[char] == total:
            count += 1
    if count == len(ransomNote):
        return True
    return False
```

```python
canConstruct(ransomNote, magazine)
```

```python
def canConstruct(ransomNote, magazine):
    note_list = list(ransomNote)
    mag_list = list(magazine)
    count = 0
    for char in note_list:
        if note_list.count(char) == mag_list.count(char):
            count += 1
    if count == len(note_list):
        return True
    return False
```

```python
def canConstruct(ransomNote, magazine):
    mag_count = Counter(magazine)
    count = 0
    for char in ransomNote:
        if char in mag_count and mag_count[char] > 1:
            mag_count[char] -= 1
            count += 1
    if count == len(ransomNote):
        return True
    return False
```
