```python
height = [1,8,6,2,5,4,8,3,7]
# height = [1, 1]
# height = [1, 2]

def maxArea(height):
    max_area = 0
    for i in range(len(height)):
        j = i + 1
        k = len(height) - 1
        while j <= k:
            min_side = min(height[i], height[j])
            max_area = max(abs(j - i) * min_side, max_area)
            j += 1
    return max_area
```

```python
def maxArea(height):
    max_area = 0
    left, right = 0, len(height) - 1
    while left <= right:
        min_side = min(height[left], height[right])
        max_area = max(abs(right - left) * min_side, max_area)
        if height[left] >= height[right]:
            right -= 1
        else:
            left += 1
    return max_area
```
