```python
def triangleNumber(nums):
    """
    :type nums: List[int]
    :rtype: int
    """
    count = 0
    nums.sort()
    for k in range(len(nums) - 1, -1, -1):
        i, j = 0, k - 1
        while i < j:
            if nums[i] + nums[j] > nums[k]:
                count += (j - i)
                j -= 1
            else:
                i += 1
    return count
```

```python
nums = [2,2,3,4]
triangleNumber(nums)
```

```python
nums = [4,2,3,4]
triangleNumber(nums)
```
