Given an integer array nums, find three numbers whose product is maximum and return the maximum product.

 

Example 1:

Input: nums = [1,2,3]
Output: 6
Example 2:

Input: nums = [1,2,3,4]
Output: 24
Example 3:

Input: nums = [-1,-2,-3]
Output: -6

```python
#，经过排序后，最大的数只可能是前两个负数和最后一个正数的乘或者是全是正数的数组的最后三个数乘积

nums = [-100,-98,-1,2,3,4]
nums.sort()
min_largest = nums[0] * nums[1] * nums[-1]
max_largest = nums[-1] * nums[-2] * nums[-3]
max(min_largest, max_largest)
```
