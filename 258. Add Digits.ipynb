```python
"""
Input: num = 38
Output: 2
Explanation: The process is
38 --> 3 + 8 --> 11
11 --> 1 + 1 --> 2 
Since 2 has only one digit, return it.
"""
```

```python
def addDigits(num):
    """
    :type num: int
    :rtype: int
    """
    while len(str(num)) > 1:
        total = 0
        for i in range(len(str(num))):
            total += int(str(num)[i])
        num = total
    return num
```

```python
def addDigits(num):
    """
    :type num: int
    :rtype: int
    """
    if num == 0:
        return num
    return 1 + (num - 1) % 9
```
