```python
nums = [1,2,3,1]
```

```python
def findPeakElement(nums):
    left, right = 0, len(nums)
    while left < right:
        mid = (left + right) // 2
        #表示更大的数字可能是mid，也可能在mid左侧
        if nums[mid] > nums[mid + 1]:
            right = mid
        #表示更大的数字一定在mid+1右侧
        elif nums[mid] < nums[mid + 1]:
            left = mid + 1
    return left
```

```python
findPeakElement(nums)
```
